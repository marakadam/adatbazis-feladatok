```python
import hashlib

algoritmusok = {"1": "sha256", "2": "md5", "3": "sha1", "4": "sha512"}


def hash_szamol(fajlnev, algoritmus):
    h = hashlib.new(algoritmus)
    with open(fajlnev, "rb") as f:
        while True:
            adat = f.read(4096)
            if not adat:
                break
            h.update(adat)
    return h.hexdigest()


def fajl_valaszt():
    try:
        from google.colab import files
        feltoltott = files.upload()
        if len(feltoltott) == 0:
            return None
        return list(feltoltott.keys())[0]
    except ImportError:
        return input("Fájl neve: ")


def algoritmus_valaszt():
    print("1 - SHA256")
    print("2 - MD5")
    print("3 - SHA1")
    print("4 - SHA512")
    valasz = input("Algoritmus: ")
    if valasz in algoritmusok:
        return algoritmusok[valasz]
    return "sha256"


while True:
    print("\n1 - Hash készítése")
    print("2 - Hash ellenőrzése")
    print("0 - Kilépés")
    menu = input("Választás: ")

    if menu == "1":
        fajl = fajl_valaszt()
        if fajl:
            alg = algoritmus_valaszt()
            print(alg, "hash:", hash_szamol(fajl, alg))

    elif menu == "2":
        fajl = fajl_valaszt()
        if fajl:
            alg = algoritmus_valaszt()
            megadott = input("Megadott hash: ").strip().lower()
            szamolt = hash_szamol(fajl, alg)
            print("Számolt hash:", szamolt)
            if szamolt == megadott:
                print("A hash egyezik, a fájl rendben van.")
            else:
                print("A hash NEM egyezik!")

    elif menu == "0":
        break
```
